**7. どんなデータが処理され、モジュールの関数やメソッドからどんなデータが返されますか？**

**入力データ**:
数値的特徴量（Recency、Discount、Profit、Quantity、Sales）およびバイナリターゲット（Churn_Flag）。

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load and clean data
monthly_sales = pd.read_csv('DATA_Orders.csv')
monthly_sales['Year Feed Parameter'] = monthly_sales['Year Feed Parameter'].astype(float).round().astype(int).astype(str)

# Group sales by year and state
grouped = monthly_sales.groupby(['Year Feed Parameter', 'State'])['Sales'].sum().unstack()

# Prepare x (years), and y1 (New York), y2 (California)
x = list(grouped.index)  # e.g., ['2022', '2023', '2024']
y1 = grouped['New York'].reindex(x).tolist()
y2 = grouped['California'].reindex(x).tolist()

# Bar positions
x_pos = np.arange(len(x))
width = 0.35

# Plot
plt.bar(x_pos - width/2, y1, width, color='green')
plt.bar(x_pos + width/2, y2, width, color='red')

plt.xticks(x_pos, x)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# Load the Excel file
df = pd.read_excel('/content/Final_Data.xlsx')

# Quick peek at data
print(df.head())

# Create churn flag based on Recency > 90 days
df['Churn_Flag'] = (df['Recency'] > 90).astype(int)

# Check churn distribution
print(df['Churn_Flag'].value_counts())

# Plot Recency distribution
sns.histplot(df['Recency'], bins=30)
plt.title('Distribution of Customer Recency')
plt.xlabel('Recency (days since last purchase)')
plt.show()

# Prepare features and target
X = df[['Recency', 'Discount', 'Profit', 'Quantity', 'Sales']]
y = df['Churn_Flag']

# Split data for training and testing
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.3)

# Train logistic regression model
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# Predict on test set
y_pred = model.predict(X_test)

# Show classification report
print(classification_report(y_test, y_pred))
```

ヒストグラムは、顧客が最後に購入した日からの経過日数（つまり、recency）を基に、顧客の最新購入状況を可視化しています。分布は強い右偏りを示しており、ほとんどの顧客は非常に最近、最初の100日以内に購入しています。Recency が増加するにつれて（つまり、購入からの日数が多くなるにつれて）、顧客の数は急激に減少します。600～1000日以上購入していない顧客が少数おり、これらの顧客は恐らく休眠顧客または離脱した顧客です。

---

The histogram visualizes how recently customers made their last purchase, measured in days since last purchase (i.e., recency). The distribution is heavily right-skewed. Most customers made purchases very recently, within the first 100 days. As recency increases (i.e., more days since last purchase), the number of customers sharply declines. A small number of customers haven't purchased in over 600–1000 days — these are likely churned or dormant.
